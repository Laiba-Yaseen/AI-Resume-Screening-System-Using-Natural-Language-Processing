Step#2: Import Libraries

```
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

Step3: Enter Job Description and Resume

```
job_description = """
We are looking for a Machine Learning Engineer with skills in Python,
Machine Learning, Deep Learning, TensorFlow, NLP, Data Analysis,
Computer Vision, and Data Visualization.
"""

resume = """
AI Engineer with experience in Python, Machine Learning, Deep Learning,
TensorFlow, NLP, Data Science projects, and data analysis.
"""
```

Step#4: Create Resume Matching Function

```
def calculate_resume_score(job, resume):

    # Combine both texts
    documents = [job, resume]

    # Convert text into numerical vectors
    tfidf = TfidfVectorizer()
    tfidf_matrix = tfidf.fit_transform(documents)

    # Calculate similarity
    similarity = cosine_similarity(
        tfidf_matrix[0:1],
        tfidf_matrix[1:2]
    )

    # Convert to percentage
    score = similarity[0][0] * 100

    return round(score, 2)
```

Step#5: Calculate Match Score

```
score = calculate_resume_score(job_description, resume)

print("========== AI Resume Screening Result ==========")
print(f"Resume Matching Score: {score}%")
```

Step#6: AI Decision

```
if score >= 80:
    print("🟢 Excellent Match! Candidate is highly suitable.")

elif score >= 50:
    print("🟡 Average Match! Candidate partially meets requirements.")

else:
    print("🔴 Low Match! Candidate needs more relevant skills.")
```
